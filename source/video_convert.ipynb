```python
# Convert videos to mp3. Running this is not needed when mp3s already available for transcription.

import subprocess
import os
import glob

def convert_mp4_to_mp3(input_file, output_file):
    ffmpeg_cmd = [
        'ffmpeg',
        '-i', input_file,
        '-vn',
        '-acodec', 'libmp3lame',
        '-b:a', '192k',
        output_file
    ]
    subprocess.run(ffmpeg_cmd, check=True)

golden_set_dir = "../golden_set"

for folder in sorted(os.listdir(golden_set_dir)):
    folder_path = os.path.join(golden_set_dir, folder)
    if not os.path.isdir(folder_path):
        continue
    mp4_files = sorted(glob.glob(os.path.join(folder_path, "*.mp4")))
    if not mp4_files:
        continue
    input_file = mp4_files[0]
    output_file = os.path.splitext(input_file)[0] + ".mp3"
    convert_mp4_to_mp3(input_file, output_file)
```
